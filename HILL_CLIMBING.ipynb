```python
import random

N = 4

def cost(board):
    """
    Cost function: counts number of attacking pairs of queens.
    board[i] = row position of queen in column i.
    """
    conflicts = 0
    for i in range(N):
        for j in range(i+1, N):
            # Same row or same diagonal
            if board[i] == board[j] or abs(board[i]-board[j]) == abs(i-j):
                conflicts += 1
    return conflicts

def print_board(board):
    """Visualize the board with Q and ."""
    for row in range(N):
        line = ""
        for col in range(N):
            if board[col] == row:
                line += "Q "
            else:
                line += ". "
        print(line)
    print()

def hill_climb():
    board = [random.randint(0, N-1) for _ in range(N)]
    print("Initial board:")
    print_board(board)
    print("Initial cost:", cost(board))

    while True:
        current_cost = cost(board)
        if current_cost == 0:
            print("Goal reached!")
            return board

        # Generate neighbors
        neighbors = []
        for col in range(N):
            for row in range(N):
                if row != board[col]:
                    new_board = board[:]
                    new_board[col] = row
                    neighbors.append((cost(new_board), new_board))

        # Sort by cost
        neighbors.sort(key=lambda x: x[0])
        best_cost, best_board = neighbors[0]

        print("Best neighbor cost:", best_cost)
        print_board(best_board)

        if best_cost >= current_cost:
            print("Stuck at local minimum with cost:", current_cost)
            return board

        board = best_board

solution = hill_climb()
print("Final solution:", solution)
print_board(solution)
```
